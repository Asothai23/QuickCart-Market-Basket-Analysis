```python
from itertools import combinations

# Convert itemsets into sets
support_dict = {}

for _, row in df.iterrows():
    items = eval(row["itemsets"], {"__builtins__": {}}, {"frozenset": frozenset})
    support_dict[frozenset(items)] = row["support"]

rules = []

for itemset, itemset_support in support_dict.items():
    if len(itemset) >= 2:
        items = list(itemset)

        for r in range(1, len(items)):
            for antecedent_tuple in combinations(items, r):
                antecedent = frozenset(antecedent_tuple)
                consequent = itemset - antecedent

                if antecedent in support_dict and consequent in support_dict:
                    confidence = itemset_support / support_dict[antecedent]
                    lift = confidence / support_dict[consequent]

                    rules.append({
                        "antecedent": ", ".join(sorted(antecedent)),
                        "consequent": ", ".join(sorted(consequent)),
                        "support": itemset_support,
                        "confidence": confidence,
                        "lift": lift
                    })

rules_df = pd.DataFrame(rules)

print("Association Rules:", len(rules_df))
rules_df.head(10)
```

```python
patterns = [
    ("Toned Milk 1L", "White Bread 400g"),
    ("Basmati Rice 5kg", "Toor Dal 1kg"),
    ("Potato Chips 90g", "Cola 750ml"),
    ("Ghee 500ml", "Poha 500g"),
    ("Instant Coffee 100g", "Choco Cookies 150g"),
    ("Hand Sanitizer 100ml", "Face Wash 100ml")
]

for a, b in patterns:
    result = rules_df[
        (rules_df["antecedent"].str.contains(a, regex=False)) &
        (rules_df["consequent"].str.contains(b, regex=False))
    ]
    print("\n", a, "->", b)
    print(result)
```

```python
useful_rules = rules_df[
    (rules_df["support"] > 0.05) &
    (rules_df["lift"] > 2)
]

print("Useful Rules:", len(useful_rules))
useful_rules[["antecedent", "consequent", "support", "confidence", "lift"]].head(20)
```
